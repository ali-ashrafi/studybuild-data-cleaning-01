### String Column Quality Assessment

```python
# Text columns

print(f"Number of text columns: {len(text_columns)}")

for column in text_columns:
    print(f"- {column}")
```

```python
# Text_Quality_Summary

text_quality_summary = []

for column in text_columns:
    # تضمین تایپ رشته‌ای
    text_series = df[column].astype("string")

    # ۱. شروع با هرگونه فضای خالی (اسپیس، تب، اینتر)
    leading_space_count = int(text_series.str.contains(r"^\s", regex=True, na=False).sum())

    # ۲. پایان با هرگونه فضای خالی
    trailing_space_count = int(text_series.str.contains(r"\s$", regex=True, na=False).sum())

    # ۳. فاقد محتوای معنادار (صرفاً فاصله‌ خالی یا پوچ)
    whitespace_only_count = int(text_series.str.strip().eq("").sum())

    text_quality_summary.append({
        "column_name": column,
        "leading_space_count": leading_space_count,
        "trailing_space_count": trailing_space_count,
        "whitespace_only_count": whitespace_only_count,})

text_quality_summary_df = pd.DataFrame(text_quality_summary)
display(text_quality_summary_df)


text_columns_with_issues = text_quality_summary_df.loc[(text_quality_summary_df[[
"leading_space_count","trailing_space_count","whitespace_only_count"]].sum(axis=1) > 0)].copy()

if text_columns_with_issues.empty:
    print("No whitespace or blank-string issues were detected.")
else:
    display(text_columns_with_issues)


total_text_issues = int(text_quality_summary_df[["leading_space_count","trailing_space_count","whitespace_only_count"]].sum().sum())
print(f"Total detected text-format issues: {total_text_issues}")    
```

```python
# 

def find_non_numeric(df, columns=numeric_columns):
    if columns is None:
        columns = df.columns
    
    results = {}
    
    for col in columns:
        converted = pd.to_numeric(df[col], errors='coerce')
        
        new_na = converted.isna()        # بعد از تبدیل NaN شده
        original_na = df[col].isna()     # از اول NaN بوده
        
        problematic_mask = new_na & ~original_na  # فقط اون‌هایی که ما باعث NaN شدیم
        
        if problematic_mask.any():
            results[col] = df[col][problematic_mask]
    
    return results


# استفاده:
non_numeric = find_non_numeric(df)

for col, values in non_numeric.items():
    print(f"\nستون '{col}':")
    print(f"  تعداد مقادیر غیر عددی: {len(values)}")
    print(f"  مقادیر منحصر به فرد: {values.unique()}")
    print(f"  ایندکس‌ها: {values.index.tolist()}")


non_numeric = find_non_numeric(df, columns=numeric_columns)
non_numeric    
```

```python
# ۱. تعریف قوانین خطا (همون دیکشنری قبلی)
validation_rules = {
    'age_invalid': (df['age'].le(0) | df['age'].gt(120)),
    'spending_invalid': (df['total_spending'].gt(15000)),
    'purchase_invalid': (df['purchase_count'].lt(0))
}

# ۲. ساخت دیتافریمِ پرچم‌های خطا
validation_df = pd.DataFrame(validation_rules)

# ۳. استخراج رکوردهای دارای مشکل از دیتافریم اصلی (df)
# متد any(axis=1) چک میکنه که آیا در هر سطر، حتی یکی از ستون‌ها True (خطا) هست یا نه
error_rows = df[validation_df.any(axis=1)]

# ۴. نمایش تعداد و خودِ رکوردها
print(f"تعداد رکوردهای دارای مشکل: {len(error_rows)}")
print("\nنمایش رکوردهای مشکل‌دار:")
display(error_rows)  # اگر توی ژوپیتر هستی از display استفاده کن، وگرنه print(error_rows)


# این کد بهت نشون میده در رکوردهای خراب، کدوم ستون‌ها True شدن
print(validation_df[validation_df.any(axis=1)].query("age_invalid | spending_invalid | purchase_invalid"))
```
